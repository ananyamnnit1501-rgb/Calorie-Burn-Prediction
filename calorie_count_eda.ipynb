```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/calories.csv')
```

```python
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

```python
sns.boxplot(x=df['Calories'])

plt.title('Boxplot of Calories')
plt.show()
```

```python
Q1 = df['Calories'].quantile(0.25)
Q3 = df['Calories'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

df = df[
    (df['Calories'] >= lower_limit) &
    (df['Calories'] <= upper_limit)
]
```

```python
sns.boxplot(x=df['Calories'])

plt.title('Boxplot of Calories')
plt.show()
```

```python


# statistics
min_val = df['Calories'].min()
q1 = df['Calories'].quantile(0.25)
median_val = df['Calories'].median()
q3 = df['Calories'].quantile(0.75)
max_val = df['Calories'].max()

# plot
plt.figure(figsize=(12,2))

sns.boxplot(x=df['Calories'])

# labels
plt.text(min_val, 0.25, f'Min\n{min_val:.2f}')
plt.text(q1, 0.25, f'Q1\n{q1:.2f}')
plt.text(median_val, 0.25, f'Median\n{median_val:.2f}')
plt.text(q3, 0.25, f'Q3\n{q3:.2f}')
plt.text(max_val, 0.25, f'Max\n{max_val:.2f}')

plt.title('Boxplot of Calories')

plt.show()
```

```python
num_cols = ['Age', 'Height', 'Weight', 'Duration',
            'Heart_Rate', 'Body_Temp', 'Calories']

for col in num_cols:
    plt.figure(figsize=(6,4))
    sns.histplot(df[col], kde=True)
    plt.title(f'Distribution of {col}')
    plt.show()
```

```python

num_cols = ['Age', 'Height', 'Weight',
            'Duration', 'Heart_Rate',
            'Body_Temp']

for col in num_cols:

    plt.figure(figsize=(6,4))

    sns.scatterplot(
        x=df[col],
        y=df['Calories']
    )

    plt.xlabel(col)
    plt.ylabel('Calories')

    plt.title(f'Calories vs {col}')

    plt.show()
```

```python
sns.countplot(x='Gender', data=df)
plt.title('Gender Distribution')
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('/content/calories.csv')
```

```python
sns.pairplot(df[['Age','Weight','Duration','Heart_Rate',
                 'Body_Temp','Calories']], diag_kind='kde')
plt.show()
```

```python
plt.figure(figsize=(10,7))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Matrix')

plt.show()
```

```python
Q1 = df['Calories'].quantile(0.25)
Q3 = df['Calories'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

df = df[
    (df['Calories'] >= lower_limit) &
    (df['Calories'] <= upper_limit)
]
```
