```python
# ============================================================
# TRAIN REVERSE MODEL
# Calories -> Workout Duration
# ============================================================

import pandas as pd
import joblib

from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split

# ============================================================
# LOAD DATASET
# ============================================================

df = pd.read_csv("/content/encoded_calories.csv")

# ============================================================
# FEATURE ENGINEERING
# ============================================================

df['BMI'] = df['Weight'] / ((df['Height'] / 100) ** 2)

df['Exercise_Intensity'] = (
    df['Heart_Rate'] * df['Duration']
)

df['Heat_Stress'] = (
    df['Body_Temp'] * df['Duration']
)

# ============================================================
# FEATURES & TARGET
# ============================================================

X_reverse = df[[
    'Calories',
    'Gender',
    'Age',
    'Height',
    'Weight',
    'BMI'
]]

y_reverse = df['Duration']

# ============================================================
# SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X_reverse,
    y_reverse,
    test_size=0.2,
    random_state=42
)

# ============================================================
# TRAIN MODEL
# ============================================================

reverse_model = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)

reverse_model.fit(X_train, y_train)

# ============================================================
# SAVE MODEL
# ============================================================

joblib.dump(
    reverse_model,
    "reverse_duration_model.pkl"
)

print("✅ Reverse duration model saved!")
```
